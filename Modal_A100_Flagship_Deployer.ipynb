```python
#@title 🚀 步驟一：貼上 Modal Token 並點 `▶` 一鍵部署你的永久專屬 A100 頂配網站！
MODAL_TOKEN_ID = "" #@param {type:"string"}
MODAL_TOKEN_SECRET = "" #@param {type:"string"}
PREFETCH_MODELS_ON_CPU_FIRST = True #@param {type:"boolean"}

import os
import subprocess
import sys

tid = MODAL_TOKEN_ID.strip()
tsec = MODAL_TOKEN_SECRET.strip()
if not tid or not tsec:
    print("❌ 請先在上方欄位填入 MODAL_TOKEN_ID (ak-...) 與 MODAL_TOKEN_SECRET (as-...) 再按 ▶ 執行！")
    raise SystemExit(0)

os.environ["MODAL_TOKEN_ID"] = tid
os.environ["MODAL_TOKEN_SECRET"] = tsec

print("📦 [1/3] 正在安裝 Modal 雲端調度器與同步最新 A100 四階段工作室程式碼（約 15 秒）...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "modal>=1.6.0"], check=True)

repo_dir = "/content/free-suno-mv-studio"
if not os.path.exists(repo_dir):
    subprocess.run(["git", "clone", "--quiet", "https://github.com/WilliamPengWp/free-suno-mv-studio.git", repo_dir], check=True)
else:
    subprocess.run(["git", "-C", repo_dir, "pull", "--quiet"], check=True)

if PREFETCH_MODELS_ON_CPU_FIRST:
    print("\n⚡ [2/3] 正在使用純 CPU 高速預載 ACE-Step 1.5 XL 與 Wan2.2-S2V-14B 權重至你的雲端硬碟（零 GPU 費用，約需數分鐘，終身只需跑一次！）...")
    subprocess.run(["modal", "run", "modal_a100_studio.py", "--prefetch-only"], cwd=repo_dir, check=True)
else:
    print("\n⏭️ [2/3] 已略過預先下載（首次在網頁點生成時也會自動用純 CPU 下載權重，不燒 GPU！）")

print("\n🚀 [3/3] 正在部署你的專屬 Serverless A100 頂配網站（閒置 $0.00/秒，內建 $24 硬熔斷鎖）...")
subprocess.run(["modal", "deploy", "modal_a100_studio.py"], cwd=repo_dir, check=True)
print("\n🎉 部署成功！請直接點擊上方印出的 `https://...-web-studio.modal.run` 網址，即可在手機或電腦隨時創作超越 Suno 的真人動態演唱 MV！")
```

```python
#@title 🎬 步驟二（選用）：免開網頁，直接在這裡一鍵產出《醉月長安（豬八戒煙嗓炸場版）》A100 真人動態 MV 並下載！
PRESET_CHOICE = "jiayibingding_s2v" #@param ["jiayibingding_s2v", "kuangge_nu_metal"]
COMPOSER_ENGINE = "ace_xl" #@param ["ace_xl", "yue2"]
VOCAL_MASTER_MODE = "stem_tube_master" #@param ["stem_tube_master", "seed_vc_mit_husky"]
VIDEO_S2V_MODE = "smart_hybrid_s2v" #@param ["smart_hybrid_s2v", "full_s2v"]
MONTHLY_SAFE_CAP_USD = 24.0 #@param {type:"slider", min:5.0, max:27.0, step:0.5}

import os
import subprocess

if not os.environ.get("MODAL_TOKEN_ID"):
    print("❌ 請先執行上方『步驟一』設定 Token！")
    raise SystemExit(0)

repo_dir = "/content/free-suno-mv-studio"
out_mp4 = f"/content/{PRESET_CHOICE}_A100_1080p.mp4"
out_mp3 = f"/content/{PRESET_CHOICE}_A100_320k.mp3"

subprocess.run([
    "modal", "run", "modal_a100_studio.py",
    "--preset", PRESET_CHOICE,
    "--composer", COMPOSER_ENGINE,
    "--vocal-mode", VOCAL_MASTER_MODE,
    "--video-mode", VIDEO_S2V_MODE,
    "--cap-usd", str(MONTHLY_SAFE_CAP_USD),
    "--out-mp4", out_mp4,
    "--out-mp3", out_mp3,
], cwd=repo_dir, check=True)

try:
    from google.colab import files
    files.download(out_mp4)
except Exception:
    print(f"✅ 檔案已產生：{out_mp4} 與 {out_mp3}")
```
